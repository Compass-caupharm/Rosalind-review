```python
protein = input()

# 각 아미노산을 암호화하는 codon 개수
codon = {
    'F': 2, 'L': 6, 'S': 6, 'Y': 2,
    'C': 2, 'W': 1, 'P': 4, 'H': 2,
    'Q': 2, 'R': 6, 'I': 3, 'M': 1,
    'T': 4, 'N': 2, 'K': 2, 'V': 4,
    'A': 4, 'D': 2, 'E': 2, 'G': 4
}

result = 3

for aa in protein:
    result = result * codon[aa] % 1000000

print(result)
```
