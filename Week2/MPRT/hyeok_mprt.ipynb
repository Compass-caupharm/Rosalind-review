```python
from urllib.request import urlopen
```

```python
with open("rosalind_mprt.txt") as f:
    id들 = f.read().split()
```

```python
for id in id들:
    url = "https://rest.uniprot.org/uniprotkb/" + id.split("_")[0] + ".fasta"
    fasta = urlopen(url).read().decode()
    seq = "".join(fasta.split("\n")[1:])

    result = []
    for i in range(len(seq) - 3):
        if seq[i] == "N" and seq[i+1] != "P" and seq[i+2] in "ST" and seq[i+3] != "P":
            result.append(str(i + 1))
        
    if len(result) > 0:
        print(id)
        print(" ".join(result))
```
