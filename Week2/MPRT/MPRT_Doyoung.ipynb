```python
from urllib.request import urlopen
import re

def get_uniprot_fasta(uniprot_id):
    
    uniprot_id = uniprot_id.split("_")[0]

    with urlopen(f"http://www.uniprot.org/uniprot/{uniprot_id}.fasta") as response:
        data = response.read()
    text = data.decode("utf-8")

    return text

def get_uniprot_sequence(uniprot_id):
    fasta = get_uniprot_fasta(uniprot_id)
    lines = fasta.splitlines()
    sequence = "".join(lines[1:])
    return sequence

def find_motif_positions(sequence):

    motif = "N[^P][ST][^P]"
    positions = []

    for i in range(len(sequence) - 3):
        if re.match(motif, sequence[i:i+4]):
            positions.append(i + 1)

    return positions

def main():
    with open("rosalind_mprt.txt") as f:
        for line in f:
            uniprot_id = line.strip()
            sequence = get_uniprot_sequence(uniprot_id)
            positions = find_motif_positions(sequence)
            
            if positions:
                print(uniprot_id)
                print(" ".join(map(str, positions)))
```

```python
if __name__ == "__main__":
    main()
```
