```python
PATH = "/Users/absolute0kelvin/Documents/Rosalind_solution/rosalind_splc.txt"

with open(PATH, "r") as file:
    content = file.read()

count = 0
full_sequence = ""
intron_sequence = []

for line in content.splitlines():
    if line.startswith(">"):
        count += 1
    else:
        if count == 1:
            full_sequence += line.strip()
        else:
            intron_sequence.append(line.strip())

print (full_sequence)
print (intron_sequence)
```

```python
spliced_sequence = full_sequence

for intron in intron_sequence:
    spliced_sequence = spliced_sequence.replace(intron, "")

codon = spliced_sequence.replace("T", "U")
print(codon)
```

```python
dict_codon_to_amino_acid = {
    "UUU": "F", "UUC": "F", "UUA": "L", "UUG": "L",
    "UCU": "S", "UCC": "S", "UCA": "S", "UCG": "S",
    "UAU": "Y", "UAC": "Y", "UAA": "*", "UAG": "*",
    "UGU": "C", "UGC": "C", "UGA": "*", "UGG": "W",
    "CUU": "L", "CUC": "L", "CUA": "L", "CUG": "L",
    "CCU": "P", "CCC": "P", "CCA": "P", "CCG": "P",
    "CAU": "H", "CAC": "H", "CAA": "Q", "CAG": "Q",
    "CGU": "R", "CGC": "R", "CGA": "R", "CGG": "R",
    "AUU": "I", "AUC": "I", "AUA": "I", "AUG": "M",
    "ACU": "T", "ACC": "T", "ACA": "T", "ACG": "T",
    "AAU": "N", "AAC": "N", "AAA": "K", "AAG": "K",
    "AGU": "S", "AGC": "S", "AGA": "R", "AGG": "R",
    "GUU": "V", "GUC": "V", "GUA": "V", "GUG": "V",
    "GCU": "A", "GCC": "A", "GCA": "A", "GCG": "A",
    "GAU": "D", "GAC": "D", "GAA": "E", "GAG": "E",
    "GGU": "G", "GGC": "G", "GGA": "G", "GGG": "G",
}

aa_string = ""

for i in range(0, len(codon) - 2, 3):
    
    amino_acid = codon[i:i+3]

    if dict_codon_to_amino_acid.get(amino_acid) == "*":
        break
    
    aa_string += dict_codon_to_amino_acid.get(amino_acid)

print(aa_string)
```
