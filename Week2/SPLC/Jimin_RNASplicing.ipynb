```python
#코돈테이블 만들기
AMINO_TO_CODONS = {
    "F": ["UUU", "UUC"],
    "L": ["UUA", "UUG", "CUU", "CUC", "CUA", "CUG"],
    "I": ["AUU", "AUC", "AUA"],
    "M": ["AUG"],
    "V": ["GUU", "GUC", "GUA", "GUG"],
    "S": ["UCU", "UCC", "UCA", "UCG", "AGU", "AGC"],
    "P": ["CCU", "CCC", "CCA", "CCG"],
    "T": ["ACU", "ACC", "ACA", "ACG"],
    "A": ["GCU", "GCC", "GCA", "GCG"],
    "Y": ["UAU", "UAC"],
    "H": ["CAU", "CAC"],
    "Q": ["CAA", "CAG"],
    "N": ["AAU", "AAC"],
    "K": ["AAA", "AAG"],
    "D": ["GAU", "GAC"],
    "E": ["GAA", "GAG"],
    "C": ["UGU", "UGC"],
    "W": ["UGG"],
    "R": ["CGU", "CGC", "CGA", "CGG", "AGA", "AGG"],
    "G": ["GGU", "GGC", "GGA", "GGG"],
    "Stop": ["UAA", "UAG", "UGA"],
}


CODON_TABLE = {}                              #실제 코드에서 mRNA를 번역할 때는 코돈을 보고 아미노산을 찾아야 하므로 "UUU": "F", "UUC": "F" 형태(CODON_TABLE)가 필요

for aa, codons in AMINO_TO_CODONS.items():    # 1. 아미노산(aa)과 코돈 리스트(codons)를 하나씩 꺼냄
    for codon in codons:                      # 2. 리스트 안에 있는 개별 코돈(codon)을 하나씩 꺼냄
        CODON_TABLE[codon] = aa               # 3. 코돈을 키(Key)로, 아미노산을 값(Value)으로 저장
```

```python
# FASTA 파일을 읽고, 이름(>Rosalind_10 등)은 버리고 순수 DNA 서열 문자열들만 순서대로 리스트에 담아 반환하는 함수
filepath="rosalind_splc.txt"

def read_fasta(filepath):
    seqs = []
    with open(filepath) as f:
        for line in f:
            line = line.strip()
            if line.startswith(">"):
                seqs.append("")   # ">"를 만날 때마다 리스트 맨 끝에 새 빈 문자열("") 자리를 하나 추가
            elif line:
                seqs[-1] += line  # 가장 마지막에 추가된 서열 칸에 이어 붙이기
    return seqs
```

```python
#RNA Splicing 함수
def rna_splicing(filepath):
    seqs = read_fasta(filepath)
    dna = seqs[0]                    # 첫 번째 서열은 dna
    introns = seqs[1:]               # 나머지는 모두 introns 리스트로 분리

    # 1. 인트론 제거
    for intron in introns:
        dna = dna.replace(intron, "")

    # 2. 전사 (DNA -> mRNA)
    rna = dna.replace("T", "U")

    # 3. 번역 (mRNA -> 단백질)
    protein = ""
    for i in range(0, len(rna) - 2, 3):
        amino_acid = CODON_TABLE.get(rna[i : i + 3], "")
        if amino_acid == "Stop":
            break
        protein += amino_acid

    return protein

rna_splicing(filepath)
```
