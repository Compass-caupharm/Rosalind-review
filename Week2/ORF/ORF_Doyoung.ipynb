```python
with open ("/Users/absolute0kelvin/Documents/Rosalind_solution/rosalind_orf.txt", "r") as file:
    content = file.read()

dna_sequence = ""

for line in content.splitlines():
    if line.startswith(">"):
        pass
    else:
        dna_sequence += line.strip()


print(dna_sequence)

reverse_complement_sequence = dna_sequence[::-1].translate(str.maketrans("ACGT", "TGCA"))

print(reverse_complement_sequence)
```

```python
dict_codon_to_amino_acid = {
    "TTT": "F", "TTC": "F", "TTA": "L", "TTG": "L",
    "CTT": "L", "CTC": "L", "CTA": "L", "CTG": "L",
    "ATT": "I", "ATC": "I", "ATA": "I", "ATG": "M",
    "GTT": "V", "GTC": "V", "GTA": "V", "GTG": "V",
    "TCT": "S", "TCC": "S", "TCA": "S", "TCG": "S",
    "CCT": "P", "CCC": "P", "CCA": "P", "CCG": "P",
    "ACT": "T", "ACC": "T", "ACA": "T", "ACG": "T",
    "GCT": "A", "GCC": "A", "GCA": "A", "GCG": "A",
    "TAT": "Y", "TAC": "Y", "TAA": "*", "TAG": "*",
    "CAT": "H", "CAC": "H", "CAA": "Q", "CAG": "Q",
    "AAT": "N", "AAC": "N", "AAA": "K", "AAG": "K",
    "GAT": "D", "GAC": "D", "GAA": "E", "GAG": "E",
    "TGT": "C", "TGC": "C", "TGA": "*", "TGG": "W",
    "CGT": "R", "CGC": "R", "CGA": "R", "CGG": "R",
    "AGT": "S", "AGC": "S", "AGA": "R", "AGG": "R",
    "GGT": "G", "GGC": "G", "GGA": "G", "GGG": "G"
}

def find_orfs(sequence):
    orfs = []
    for i in range(len(sequence) - 2):
        codon = sequence[i:i+3]
        if codon == "ATG":
            orf = ""
            for j in range(i, len(sequence) - 2, 3):
                codon = sequence[j:j+3]
                amino_acid = dict_codon_to_amino_acid.get(codon, "")
                if amino_acid == "*":
                    orfs.append(orf)
                    break
                orf += amino_acid
    return orfs

output = set(find_orfs(dna_sequence) + find_orfs(reverse_complement_sequence))

for protein in output:
    print(protein)
```
