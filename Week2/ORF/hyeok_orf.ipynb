```python
from Bio import SeqIO

record = SeqIO.read("example.fasta", "fasta")
dna = record.seq                  
reverse_dna = dna.reverse_complement()
```

```python
amino = set()

for i in range(len(dna) - 2):
    if dna[i:i+3] == "ATG":
        orf = dna[i:]
        protein = orf.translate()
        if "*" in protein:
            stop = protein.index("*")
            amino.add(protein[:stop])
```

```python
for i in range(len(reverse_dna) - 2):
    if reverse_dna[i:i+3] == "ATG":
        orf = reverse_dna[i:]
        protein = orf.translate()
        if "*" in protein:
            stop = protein.index("*")
            amino.add(protein[:stop])
```

```python
for k in amino:
    print(k)
```
