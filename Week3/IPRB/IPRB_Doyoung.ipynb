```python
k = int(input("put AA"))
m = int(input("put Aa"))
n = int(input("put aa"))

total = k + m + n

# calculate probabilities to get recessive phenotype

Aa_Aa = (m / total * 1/2) * ((m-1) / (total - 1) * 1/2)
Aa_aa = (m / total * 1/2) * (n / (total - 1) * 1)
aa_Aa = (n / total * 1) * (m / (total-1) * 1/2)
aa_aa = (n / total * 1) * ((n-1) / (total - 1) * 1)

recessive_phenotype = Aa_Aa + Aa_aa + aa_Aa + aa_aa
dominant_phenotype = 1 - recessive_phenotype

print(dominant_phenotype)
```
