```python
# dictionary로 우동과 이형과 열동 각각의 확률을 적고,, 표현
def 멘델(k,m,n):
    mendel = {"우동":k,"이형":m,"열동":n}
    A = k+m+n
    전체확률 = 0
    for key1 in mendel.keys():
        copy = mendel.copy()
        X = (copy[key1])/A
        if key1 == "우동":
            copy["우동"] += -1
        if key1 == "이형":
            copy["이형"] += -1
        if key1 == "열동":
            copy["열동"] += -1
        for key2 in mendel.keys():
             Y = (copy[key2])/(A-1)
             if key1 == "이형" and key2 == "이형":
                 Y *= 3/4
             if key1 == "이형" and key2 == "열동":
                 Y *= 1/2
             if key1 == "열동" and key2 == "이형":
                 Y *= 1/2
             if key1 == "열동" and key2 =="열동":
                 Y *= 0
             전체확률 += X * Y
    return 전체확률
print(멘델(19,26,16))
```

```python
# 확률보다 경우의수로 푸는 게 훨씬 간단함
def 멘델(k, m, n):
    t = k + m + n
    return (k*(k-1) + 2*k*m + 2*k*n + 0.75*m*(m-1) + m*n) / (t*(t-1))

print(멘델(999, 1000, 1234))   
```
