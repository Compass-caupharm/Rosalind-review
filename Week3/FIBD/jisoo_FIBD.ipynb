```python
n = 6    # 몇 달째까지 볼지
m = 3    # 토끼 수명

# ---- 나이별 토끼 수를 담을 리스트 만들기 ----
# ages[0] = 1살, ages[1] = 2살, ..., ages[m-1] = m살
ages = []
for i in range(m):
    ages.append(0)       # 처음엔 전부 0
ages[0] = 1              # 1달째: 새끼(1살) 1쌍으로 시작

print("1 달째:", ages, "합계:", sum(ages))

# ---- 2달째부터 n달째까지 한 달씩 넘기기 ----
for month in range(2, n + 1):

    # ① 새끼 수 = 지난달 2살 이상 토끼 수의 합
    newborn = 0
    for i in range(1, m):            # 1번 칸(2살)부터 끝까지
        newborn = newborn + ages[i]

    # ② 나이 먹기: 새 리스트에 한 칸씩 옮겨 담기
    new_ages = []
    new_ages.append(newborn)         # 1살 자리에는 새로 태어난 새끼
    for i in range(0, m - 1):        # 마지막 칸(m살)은 옮기지 않음 → 죽음
        new_ages.append(ages[i])     # i번 칸 토끼가 한 살 더 먹어서 다음 칸으로

    ages = new_ages                  # 이번 달 상태로 바꾸기

    print(month, "달째:", ages, "합계:", sum(ages))

# ---- 최종 답 ----
print(sum(ages))
```
