```python
n, m = 6, 3

ages = [1] + [0] * (m - 1)        # 1월: [0살, 1살, ..., (m-1)살] = [1, 0, ..., 0]
print(1, "월:", ages, "합계:", sum(ages))

for month in range(2, n + 1):
    newborn = sum(ages[1:])        # 규칙 ① 새끼 = 지난달 1살~(m-1)살의 합
    ages = [newborn] + ages[:-1]   # 규칙 ② 한 칸씩 이동, 규칙 ③ 마지막 칸 삭제
    print(month, "월:", ages, "합계:", sum(ages))

print(sum(ages))
```
