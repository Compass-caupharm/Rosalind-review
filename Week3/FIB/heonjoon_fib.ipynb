```python
import sys
```

```python
def rabbits(n, k):
    a, b = 1, 1                 # 1개월째, 2개월째 토끼 쌍 수
    for _ in range(n - 2):      # 3개월째부터 n개월째까지 반복
        a, b = b, b + k * a     # 이번 달 = 지난달 + k × 두 달 전
    return b
```

```python
# 샘플 데이터로 확인 (정답: 19)
print(rabbits(5, 3))
```

```python
# 데이터셋 파일로 실행
filename = "rosalind_fib.txt"   # 노트북과 같은 폴더에 둔 데이터셋 파일
n, k = map(int, open(filename).read().split())
print(rabbits(n, k))
```
