```python
1개월차 : 아무도 없다 2개월차: 성인1쌍 3개월차 성ㄴ인1상 새끼 3쌍 
4개월차 성인 4쌍 새끼 3쌍 5개월차 성인 7쌍 새끼 12쌍 6개월차 성인 17 쌍 새끼 21쌍 
7개월차 성인 41쌍 새끼 51쌍 
```

```python

def 새끼(n,k):
    새끼 = 1
    성인 = 0
    for i in range(0,n-1):
        성인, 새끼 = 성인 + 새끼, k * 성인 #이미 설정된 성인과 새끼 값을 씀, 이 두가지를 두 줄로 나눠서 작성하면 오류가 생김
    return 성인 + 새끼
print(새끼(32,3))
```
