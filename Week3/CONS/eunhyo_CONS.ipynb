```python
with open("rosalind_cons.txt", encoding="utf-8") as f:
    data = f.read()

seqs = []
for a in data.split(">")[1:]:
    lines = a.split()
    if lines:
        seq = "".join(lines[1:])
        seqs.append(seq)
n = len(seqs[0])
A, C, G, T = [0] * n, [0] * n, [0] * n, [0] * n

for l in seqs:
    for j, base in enumerate(l):
        if base == "A":
            A[j] += 1
        elif base == "C":
            C[j] += 1
        elif base == "G":
            G[j] += 1
        elif base == "T":
            T[j] += 1
consensus_chars = []
for j in range(n):
    counts = [("A", A[j]), ("C", C[j]), ("G", G[j]), ("T", T[j])]
    consensus_string, _ = max(counts, key=lambda x: x[1])
    consensus_chars.append(consensus_string)
print("".join(consensus_chars))
print("A:", *A)
print("C:", *C)
print("G:", *G)
print("T:", *T)
```
