```python
f = open("rosalind_cons.txt")
data = f.read()

seqs = []
for block in data.split(">")[1:]:
    lines = block.split()
    seq = "".join(lines[1:])
    seqs.append(seq)

A = []
C = []
G = []
T = []
consensus = ""

for j in range(len(seqs[0])):
    column = ""
    for s in seqs:
        column += s[j]

    a = column.count("A")
    c = column.count("C")
    g = column.count("G")
    t = column.count("T")

    A.append(a)
    C.append(c)
    G.append(g)
    T.append(t)

    m = max(a, c, g, t)       
    if a == m:
        consensus += "A"
    elif c == m:
        consensus += "C"
    elif g == m:
        consensus += "G"
    else:
        consensus += "T"

print(consensus)
print("A:", *A)
print("C:", *C)
print("G:", *G)
print("T:", *T)
```
