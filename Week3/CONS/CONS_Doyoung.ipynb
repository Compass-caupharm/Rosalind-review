```python
PATH = "/Users/absolute0kelvin/Documents/Rosalind_solution/rosalind_cons.txt"

with open(PATH, "r") as file:
    content = file.read()

sequence_list = []
sequence_string = ""

for line in content.splitlines():
    if line.startswith(">"):
        if sequence_string != "":
            sequence_list.append(sequence_string)
            sequence_string = ""

    else:
        sequence_string += line

sequence_list.append(sequence_string)
```

```python
base_dict = {'A':[], 'C':[], 'G':[], 'T':[]}
consensus_string = ""
name = "ACGT"

for col in zip(*sequence_list):
    base_dict['A'].append(col.count('A'))
    base_dict['C'].append(col.count('C'))
    base_dict['G'].append(col.count('G'))
    base_dict['T'].append(col.count('T'))
    values = (col.count('A'), col.count('C'), col.count('G'), col.count('T'))

    i = values.index(max(values))
    consensus_string += name[i]

print(consensus_string)

for base, matrix in base_dict.items():
    print(f"{base}" + ": " + " ".join(map(str, matrix)) )
```
