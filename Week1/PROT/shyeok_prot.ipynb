```python
#코돈 표 텍스트 파일을 따로 만들고 열어서 딕셔너리로 변환
with open("RNA codon table.txt", encoding="utf-8") as c:
    raw_amino=c.read().split()    #['UUU', 'F', ..., 'GGG', 'G']

codon_table = raw_amino[0::2]     #['UUU', 'CUU', ..., 'AGG', 'GGG']
amino_table = raw_amino[1::2]     #['F', ..., 'G']

codon_dict=dict(zip(codon_table,amino_table))

print(codon_dict)
```

```python
#파일 받아서 읽기
with open("rosalind_prot.txt", encoding="utf-8") as origin_text:
    mrna=origin_text.read().strip()     #AUGCCC...UACUUAA

#반복문 이용해서 mrna 리스트 3개씩 slicing
codons = [mrna[i:i+3] for i in range (0, len(mrna)-2, 3)]     #['AUG', 'CCC', ..., 'ACU', 'UAA']

#코돈에 대응되는 아미노산 딕셔너리에서 찾아 지정하기
prot_answer = [codon_dict.get(x) for x in codons] 
print("".join(prot_answer[:prot_answer.index("Stop")]))    #MPNR...LSCSLVT
```
