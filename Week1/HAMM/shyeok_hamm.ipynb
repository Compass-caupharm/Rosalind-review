```python
# 파일을 열어 두 줄을 s, t에 저장
with open("rosalind_hamm.txt", encoding="utf-8") as f:
    lines = f.read().splitlines()   # 줄 단위 리스트로 분리
s = lines[0]    #AAC...GAA
t = lines[1]    #AGC...ATA

# 같은 위치의 염기를 하나씩 비교
hamming_distance = 0    # 서로 다른 자리의 개수

for i in range(0, len(s)):  # 0부터 len(s)-1까지 위치 번호
    if s[i] != t[i]:    # 같은 위치의 두 염기가 다르면
        hamming_distance += 1   # 개수에다가 +1

print(hamming_distance)
```
